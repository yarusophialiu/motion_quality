```python
import pandas as pd
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
res = pd.read_csv('data.csv', delimiter  = ',')
u_rr = pd.unique(res['refresh_rate'])
u_sc = pd.unique(res['scene'])
clnames = ['bdwth', 'rr', 'sc','mss','mms','mos','sss','sms','sos','cmmsmss','cmosmss','cmmsmos']
```

```python
all_res = pd.DataFrame(columns =clnames)
for ii in u_rr:
    for jj in u_sc:
        cur_df = res[res['refresh_rate']==ii][res['scene']==jj]
        cur_df = cur_df.reset_index(drop=True)
        bdwidth = cur_df['bandwidth'][0]
        mss = np.mean(cur_df['sharpnness_score'])
        sss = np.std(cur_df['sharpnness_score'])
        mms = np.mean(cur_df['motion_score'])
        sms = np.std(cur_df['motion_score'])
        mos = np.mean(cur_df['overall_score'])
        sos = np.std(cur_df['overall_score'])
        cmmsmss = np.corrcoef(cur_df['sharpnness_score'],cur_df['motion_score'])[0,1]
        cmosmss = np.corrcoef(cur_df['overall_score'],cur_df['sharpnness_score'])[0,1]
        cmmsmos = np.corrcoef(cur_df['motion_score'],cur_df['overall_score'])[0,1]
        df_tmp = pd.DataFrame(columns =clnames, 
                            data=[[bdwidth,ii,jj,mss,mms,mos,sss,sms,sos,cmmsmss,cmosmss,cmmsmos]])

        all_res = all_res.append(df_tmp)
        all_res.reset_index(inplace= True, drop = True)
```

```python
all_res = all_res.sort_values(by=['rr'])
```

```python
metrics = ['mss','mms','mos']
metricsttl = ['Sharpness','Motion','Overall']
stdmetrics = ['sss','sms','sos']
lbls = [ 'US', '30', '60', '120','165']
for jj in u_sc:
    
    for ii in range(0,3):
        fig, ax = plt.subplots()
        
        bar_plt = ax.bar(x = np.arange(len(u_rr)), 
               height = all_res[all_res['sc']==jj][metrics[ii]], 
               yerr=all_res[all_res['sc']==jj][stdmetrics[ii]], 
               align='center', alpha=0.5, 
               ecolor='black', capsize=10)
        ax.set_xticks(np.arange(0,5))
        ax.set_xticklabels(lbls)
        ax.set_title(jj+' '+metricsttl[ii])
        ax.yaxis.grid(True)
        # Add numbers in plot
        sterr = all_res[all_res['sc']==jj][stdmetrics[ii]].tolist() 
        
        meanerr = all_res[all_res['sc']==jj][metrics[ii]].tolist() 
        #print(sterr)
        for i, rectangle in enumerate(bar_plt):
            height = rectangle.get_height()
            plt.text(rectangle.get_x() + rectangle.get_width()/2, height+sterr[i] + 0.3,
             '$\mu=$%s \n $\sigma=$%s' % (str(round(meanerr[i],3)),
                                          str(round(sterr[i],3))),
             ha='center', va='bottom')
        
        plt.show()

        
```

```python
srocc_ttl = ['cmosmss','cmmsmos']

srocc_ttl_comp = ['PLCC Overall vs Sharpness','PLCC Overll vs Motion']

for ii in range(0,2):
    
    mean_list = list()
    std_list = list()
    for jj in u_sc:
        mean_list.append(np.mean(all_res[all_res['sc']==jj][srocc_ttl[ii]]))
        std_list.append(np.std(all_res[all_res['sc']==jj][srocc_ttl[ii]]))
    

    fig, ax = plt.subplots()
    bar_plt = ax.bar(x = np.arange(len(u_sc)), 
           height = mean_list,
           yerr=std_list,
           align='center', alpha=0.5, 
           ecolor='black', capsize=10)
    ax.set_xticks(np.arange(0,3))
    ax.set_xticklabels(u_sc)
    ax.set_title(srocc_ttl_comp[ii])
    ax.yaxis.grid(True)
    
    ax.set_ylim(top=1)

    for i, rectangle in enumerate(bar_plt):
        height = rectangle.get_height()
        plt.text(rectangle.get_x() + rectangle.get_width()/2, height+std_list[i] + 0.01,
         '$\mu=$%s \n $\sigma=$%s' % (str(round(mean_list[i],3)),
                                      str(round(std_list[i],3))),
         ha='center', va='bottom')

    plt.show()
```
